```python
import numpy as np
import time
import matplotlib.pyplot as plt
from pynq import Overlay, allocate

print("--- Systolic Array Research Benchmarking Suite ---")

# ==========================================
# 1. HARDWARE & BENCHMARK PARAMS
# ==========================================
TILE = 32
BITSTREAM = "design_1.bit"

# Define the shapes to sweep: (REAL_M, REAL_K, REAL_N)
# Mix of square matrices and your specific neural network layer sizes
SHAPES_TO_TEST = [
    (32, 32, 32),
    (64, 64, 64),
    (128, 128, 128),
    (256, 256, 256),
    (512, 512, 512),
    (1024, 1024, 1024),
    (1, 784, 128),   # Your specific FC layer
    (16, 1024, 128)  # Batched FC layer
]

# Hardware Theoretical Peaks (Adjust these based on your specific FPGA/Design)
# Assuming 100 MHz clock and a 32x32 array (1024 MACs = 2048 Ops/cycle)
PEAK_GOPS = 204.8 
# Assuming DDR3 standard bandwidth on PYNQ-Z1/Z2
PEAK_BW_GBPS = 4.2 

# ==========================================
# 2. HELPER FUNCTIONS
# ==========================================
def align(dim):
    return ((dim + TILE - 1) // TILE) * TILE

def pack_to_blocked(matrix, rows, cols):
    return matrix.reshape(rows // TILE, TILE, cols // TILE, TILE) \
                 .transpose(0, 2, 1, 3).flatten()

def unpack_from_blocked(flat_array_32bit, rows, cols):
    r_tiles = rows // TILE
    c_tiles = cols // TILE
    blocked = flat_array_32bit.reshape(r_tiles, c_tiles, TILE, TILE)
    return blocked.transpose(0, 2, 1, 3).reshape(rows, cols)

def assign_reg(ip, name, value):
    if hasattr(ip.register_map, name):
        setattr(ip.register_map, name, value)
    elif hasattr(ip.register_map, name + "_1"):
        setattr(ip.register_map, name + "_1", value)
        setattr(ip.register_map, name + "_2", 0)
    else:
        raise RuntimeError(f"Register '{name}' missing from IP.")

# ==========================================
# 3. BENCHMARKING LOOP
# ==========================================
print("Loading Bitstream...")
overlay = Overlay(BITSTREAM, download=True)
systolic_ip = overlay.pipelined_layer_proc_0

results_gops = []
results_oi = []
labels = []

for shape in SHAPES_TO_TEST:
    REAL_M, REAL_K, REAL_N = shape
    M, K, N = align(REAL_M), align(REAL_K), align(REAL_N)
    
    print(f"\n[*] Benchmarking Shape: {REAL_M}x{REAL_K}x{REAL_N} (Padded: {M}x{K}x{N})")
    
    # 1 byte for int8 A and B, 4 bytes for int32 C
    bytes_transferred = (M * K * 1) + (K * N * 1) + (M * N * 4) 
    ops = 2 * M * K * N 
    operational_intensity = ops / bytes_transferred
    
    # Allocate buffers
    buf_A = allocate(shape=(M * K,), dtype=np.int8)
    buf_B = allocate(shape=(K * N,), dtype=np.int8)
    buf_C = allocate(shape=(M * N,), dtype=np.int32)
    
    try:
        # Initialize Random Data
        A_sw = np.random.randint(-128, 127, size=(M, K), dtype=np.int8)
        B_sw = np.random.randint(-128, 127, size=(K, N), dtype=np.int8)
        
        buf_A[:] = pack_to_blocked(A_sw, M, K)
        buf_B[:] = pack_to_blocked(B_sw, K, N)
        buf_A.flush()
        buf_B.flush()
        
        # Configure Hardware
        assign_reg(systolic_ip, 'mat_A', buf_A.physical_address)
        assign_reg(systolic_ip, 'mat_B', buf_B.physical_address)
        assign_reg(systolic_ip, 'mat_C', buf_C.physical_address)
        assign_reg(systolic_ip, 'num_row_tiles', M // TILE)
        assign_reg(systolic_ip, 'num_depth_tiles', K // TILE)
        assign_reg(systolic_ip, 'num_col_tiles', N // TILE)
        
        # WARMUP RUN
        systolic_ip.write(0x00, 0x01)
        while not (systolic_ip.read(0x00) & 0x02):
            pass
            
        # TIMED RUNS
        iterations = 50
        start_time = time.time()
        for _ in range(iterations):
            systolic_ip.write(0x00, 0x01)
            while not (systolic_ip.read(0x00) & 0x02):
                pass
        end_time = time.time()
        
        # Calculate Metrics
        avg_time_s = (end_time - start_time) / iterations
        gops = (ops / avg_time_s) / 1e9
        
        print(f"    Avg Time:  {avg_time_s * 1000:.4f} ms")
        print(f"    Throughtput: {gops:.2f} GOPS")
        print(f"    Op Intensity: {operational_intensity:.2f} FLOPs/Byte")
        
        results_gops.append(gops)
        results_oi.append(operational_intensity)
        labels.append(f"{REAL_M}x{REAL_K}x{REAL_N}")
        
    finally:
        # CRITICAL: Prevent CMA memory exhaustion during sweep
        buf_A.close()
        buf_B.close()
        buf_C.close()

# ==========================================
# 4. PLOTTING GRAPHS
# ==========================================
print("\nGenerating Research Graphs...")

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

# Plot 1: Throughput Bar Chart
x_pos = np.arange(len(labels))
bars = ax1.bar(x_pos, results_gops, color='#4C72B0')
ax1.set_xticks(x_pos)
ax1.set_xticklabels(labels, rotation=45, ha='right')
ax1.set_ylabel('Throughput (GOPS)')
ax1.set_title('Systolic Array Performance Scaling')
ax1.grid(axis='y', linestyle='--', alpha=0.7)

# Add value labels on top of bars
for bar in bars:
    yval = bar.get_height()
    ax1.text(bar.get_x() + bar.get_width()/2, yval + 0.5, round(yval, 1), ha='center', va='bottom')

# Plot 2: Empirical Roofline Model (Log-Log Scale)
ax2.scatter(results_oi, results_gops, color='red', zorder=5, s=100, label="Empirical Data")

# Annotate points on Roofline
for i, txt in enumerate(labels):
    ax2.annotate(txt, (results_oi[i], results_gops[i]), xytext=(5, 5), textcoords='offset points', fontsize=9)

# Draw Theoretical Rooflines
x_oi = np.logspace(np.log10(min(results_oi)*0.5), np.log10(max(results_oi)*2), 100)
y_compute_bound = np.full_like(x_oi, PEAK_GOPS)
y_memory_bound = x_oi * PEAK_BW_GBPS

# Plot roofs
ax2.plot(x_oi, y_compute_bound, 'k--', label=f'Peak Compute ({PEAK_GOPS} GOPS)')
ax2.plot(x_oi, y_memory_bound, 'b--', label=f'Peak Mem BW ({PEAK_BW_GBPS} GB/s)')

# The actual roof is the minimum of the two bounds
y_roof = np.minimum(y_compute_bound, y_memory_bound)
ax2.plot(x_oi, y_roof, 'k-', linewidth=2, alpha=0.5)

ax2.set_xscale('log')
ax2.set_yscale('log')
ax2.set_xlabel('Operational Intensity (Ops/Byte)')
ax2.set_ylabel('Performance (GOPS)')
ax2.set_title('Empirical Roofline Model')
ax2.legend()
ax2.grid(True, which="both", ls="--", alpha=0.5)

plt.tight_layout()
plt.savefig("research_benchmark.png", dpi=300)
plt.show()

print("\nBenchmarking complete! Plots saved to 'research_benchmark.png'.")
```
